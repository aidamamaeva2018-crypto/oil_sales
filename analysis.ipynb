```python
df.to_sql("oil", conn, if_exists="replace", index=False)
q1 = df["value_sales"].quantile(0.25)
q3 = df["value_sales"].quantile(0.75)
iqr = q3-q1
lower = q1 - 1.5*iqr
upper = q3 + 1.5*iqr
run(f"SELECT * FROM oil WHERE value_sales> {upper} OR value_sales < {lower} ORDER BY value_sales DESC")
```

```python
df["price_bracket"] = df["price_bracket"].astype(str).str.replace("$", "", regex=False)
```

```python
def parse_bracket(s):
    if "+" in s:
        low = s.replace("+", "").strip()
        return float(low), None
    elif "-" in s:
        parts = s.split("-")
        if len(parts) == 2:
            try:
                low = float(parts[0].strip())
                hight = float(parts[1].strip())
                return low, hight
            except: 
                return None, None
    return None, None

parsed = df["price_bracket"].apply(parse_bracket)
df["bracket_low"] = parsed.apply(lambda x:x[0])
df["bracket_hight"] = parsed.apply(lambda x:x[1])
```

```python
out = (
    (df["average_price"] < df["bracket_low"])| ((df["average_price"] > df["bracket_hight"]) & (df["bracket_hight"].notna()))
) 
anomalies = df[out]
print(f"Найдено {len(anomalies)} аномалий")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import TimeSeriesSplit
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score, mean_squared_log_error
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

df["size_num"] = df["size"].str.extract(r"(\d+\.?\d*)").astype(float)

def middle(s):
    if isinstance(s, str):
        s = s.replace("$", "").strip()
        if "-" in s:
            parts = s.split("-")
            try:
                low = float(parts[0])
                hight = float(parts[1])
                return (low + hight)/2
            except:
                return None
        elif "+" in s:
            try: return float(s.replace("+", ""))*1.2
            except: return None
    return None

df["middle"] = df["price_bracket"].apply(middle)
df["price_per_liter"] = df["average_price"]/df["size_num"]
df["quarter"] = (df["month"] - 1)//3+1
df["is_wholesale"] = df["volume_sales"] > df["volume_sales"].quantile(0.9).astype(int)

features = ["class", "size_num", "middle", "price_per_liter", "volume_sales", "month", "quarter", "is_wholesale", "year"]

x_train = df.loc[df["year"]<2024, features]
x_test = df.loc[df["year"]==2024, features]

y_train_log = np.log1p(df.loc[df["year"]<2024, "value_sales"])
y_test_log = np.log1p(df.loc[df["year"]==2024, "value_sales"])


cat_features = ["class"]
num_features = [col for col in features if col not in cat_features]

preprocessor = ColumnTransformer([
    ("num", StandardScaler(), num_features), ("cat", OneHotEncoder(drop = "first", handle_unknown = "ignore"), cat_features)
])



model_log = Pipeline([
    ("preprocessor", preprocessor), ("regressor", LinearRegression())
])

model_log.fit(x_train, y_train_log)

pred_log = model_log.predict(x_test)
pred = np.expm1(pred_log)
true = np.expm1(y_test_log)

rmse = np.sqrt(mean_squared_error(true, pred))
mae = mean_absolute_error(true, pred)
r_2 = r2_score(true, pred)
rmsle = np.sqrt(mean_squared_log_error(true, np.maximum(pred, 0)))

print(f"rmse: {rmse:.2f}")
print(f"mae: {mae:.2f}")
print(f"R^2: {r_2:.2f}")
print(f"rmsle: {rmsle:.2f}")

```

```python
import pandas as pd
import numpy as np
from xgboost import XGBRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score, mean_squared_log_error
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

cat_features = ["class"]
num_features = [col for col in features if col not in cat_features]

preprocessor = ColumnTransformer([
    ("num", StandardScaler(), num_features), ("cat", OneHotEncoder(drop = "first", handle_unknown = "ignore"), cat_features)
])

xgb = XGBRegressor(
    n_estimators=200,
    learning_rate = 0.1,
    max_depth = 6,
    random_state = 42
)

pipe = Pipeline([
    ("prep", preprocessor), ("xgb", xgb)
])
pipe.fit(x_train, y_train_log)

y_pred_log = pipe.predict(x_test)
y_pred = np.expm1(y_pred_log)
y_true = np.expm1(y_test_log)


rmse = np.sqrt(mean_squared_error(y_true, y_pred))
mae = mean_absolute_error(y_true, y_pred)
r_2 = r2_score(y_true, y_pred)
rmsle = np.sqrt(mean_squared_log_error(y_true, np.maximum(y_pred, 0)))

print(f"rmse: {rmse:.2f}")
print(f"mae: {mae:.2f}")
print(f"R^2: {r_2:.2f}")
print(f"rmsle: {rmsle:.2f}")
```
